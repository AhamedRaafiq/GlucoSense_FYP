```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Input: Specify the full file path here
file1_path = r"C:\Users\DELL\Documents\GitHub\fyp\07_Machine_Learning_Models\sai_wali_file.txt"

# Verify file exists
if not os.path.exists(file1_path):
    raise FileNotFoundError(f"File not found: {file1_path}")

df1 = pd.read_csv(file1_path, delim_whitespace=True)

print("File loaded successfully!")
print(f"Data shape: {df1.shape}")
print(f"Columns: {df1.columns.tolist()}")
```

```python
# Butterfly Curve for SRAM Cell
plt.figure(figsize=(10, 8))

# Plot the first curve: V(q) vs V(q_bar)
plt.plot(df1["V(q)"], df1["V(q_bar)"], 'b-', linewidth=2, label="Curve 1", alpha=0.7)

# Plot the second curve (inverted): V(q_bar) vs V(q) - this creates the butterfly effect
plt.plot(df1["V(q_bar)"], df1["V(q)"], 'g-', linewidth=2, label="Curve 2", alpha=0.7)

plt.xlabel("V$_Q$ (V)", fontsize=14, fontweight='bold')
plt.ylabel("V$_{\overline{Q}}$ (V)", fontsize=14, fontweight='bold')
plt.title("Butterfly curve for the SRAM cell", fontsize=16, fontweight='bold')
plt.legend(fontsize=12, loc='upper right')
plt.grid(True, alpha=0.3, linestyle='--')

# Set equal aspect ratio and appropriate limits
plt.axis('equal')
plt.xlim([0, 2])
plt.ylim([0, 2])

plt.tight_layout()
plt.show()
```

```python
# Individual plots for reference
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Curve 1: V(q) vs V(q_bar)
ax1.plot(df1["V(q)"], df1["V(q_bar)"], 'b-', linewidth=2)
ax1.set_xlabel("V(q) (V)", fontsize=12)
ax1.set_ylabel("V(q_bar) (V)", fontsize=12)
ax1.set_title("Curve 1: V(q) vs V(q_bar)", fontsize=14)
ax1.grid(True, alpha=0.3)
ax1.set_aspect('equal')

# Curve 2: V(q_bar) vs V(q) (inverted)
ax2.plot(df1["V(q_bar)"], df1["V(q)"], 'g-', linewidth=2)
ax2.set_xlabel("V(q_bar) (V)", fontsize=12)
ax2.set_ylabel("V(q) (V)", fontsize=12)
ax2.set_title("Curve 2: V(q_bar) vs V(q) (Inverted)", fontsize=14)
ax2.grid(True, alpha=0.3)
ax2.set_aspect('equal')

plt.tight_layout()
plt.show()
```
