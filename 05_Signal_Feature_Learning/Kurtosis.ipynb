```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gennorm
from ipywidgets import interact, FloatSlider

def plot_interactive_kurtosis(beta=2.0):
    """
    Plots the Generalized Normal Distribution with a slider for Beta.
    """
    x = np.linspace(-5, 5, 1000)
    
    # Calculate PDF and Kurtosis
    y = gennorm.pdf(x, beta)
    kurt_val = gennorm.stats(beta, moments='k')
    
    plt.figure(figsize=(10, 6))
    
    # Plot the line
    plt.plot(x, y, linewidth=3, color='dodgerblue', label=f'Beta: {beta:.2f}')
    
    # Fill under the curve for better visibility
    plt.fill_between(x, y, color='dodgerblue', alpha=0.1)
    
    # Set fixed limits so the axes don't jump around while sliding
    plt.ylim(0, 1.0)
    plt.xlim(-5, 5)
    
    plt.title(f'Visualizing Kurtosis\nExcess Kurtosis: {kurt_val:.2f}', fontsize=14)
    plt.xlabel('Value')
    plt.ylabel('Probability Density')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

# Create the slider
# min=0.5, max=10.0, step=0.1, starting_value=2.0
interact(plot_interactive_kurtosis, 
         beta=FloatSlider(value=2.0, min=0.5, max=10.0, step=0.1, description='Beta (Shape):'));
```
